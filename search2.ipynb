```python
from googlesearch import search
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
topic = "mental health"
suffix = f"{topic} wikipedia"

real = search(suffix, num_results=2)
links = list(real)
links[0]
```

```python
url = links[0]
try:
    page = requests.get(url)
except Exception as e:
    print(e)


soup = BeautifulSoup(page.text, 'html.parser')

for i in range(0,6):
    print(soup.find_all('p')[i].get_text())


# write to a file, then send the file to ollama api to generate 6 headlines, again use google search api 
```
